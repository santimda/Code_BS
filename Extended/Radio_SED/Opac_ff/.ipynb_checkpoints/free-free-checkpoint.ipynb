```python
import numpy as np
```

```python
# Constants
eV = 1.602176e-12  # electron-volt
c = 2.99792458e10  # speed of light in vacuum
k = 1.380658e-16   # boltzmann's constant 
qe = 4.8032068e-10 # electron charge
mec2 = 0.511e6*eV  # electron rest energy
re = 2.8e-13       # electron classical radius
sigma = 5.6705e-5  # stephan-boltzmann's constant
sigma_T = 6.652459e-25  # Thomson cross-section for e-
h = 6.6260755e-27  # planck's constant
alfa = 7.29735308e-3 # fine structure constant
M_sun_yr = 1.989e33/31556926.0 # M_sun/yr
R_sun = 6.96e10    # solar radius 
L_sun = 4e33      # solar luminosity (CGS)
AU = 1.496e13      # Astronomical Unit
pc = 3.0856e18     # parsec
kpc = 3.0856e21    # kiloparsec
mp = 1.673e-24     # proton mass
g_ad = 5./3.   # adiabatic coefficient 
m_pic2 = 134.98e6*eV # pion rest energy
mpc2 = 1836.0*mec2  # proton rest energy
Ry = 2.17987e-11   # Rydberg unit = 2.18e-18 J
mJy = 1.0e-26
```

```python
# Stellar parameters
f_w = 0.2 # volume filling factor of the winds (assumption)

# WR 142
d1 = 1.72 * kpc       # Distance to the system (Gaia DR3)
T_st1 = 4.07e4         # Sander+2019?
Rst1 = 19*R_sun      # Sander+2019?
vinf1 = 3000e5        # Sander+2019?
Mdot1 = 1.6e-5 * M_sun_yr # Sander+2019? 
mu_w1 = 4.0           # ???
Zq_w1 = 1.005         # rms ionic charge
gamma_w1 = 1.01       # Leitherer 1995, ????
T_w1 = 0.3 * T_st1    # Drew1990 for r >> Rst

# WR 114
d2 = 1.97 * kpc       # Distance to the system (Gaia DR3)
T_st2 = 79e3          # Sander+2019?
Rst2 = 2.68*R_sun     # Sander+2019?
vinf2 = 3200e5        # Sander+2019?
Mdot2 = 3.1e-5 * M_sun_yr # ??? 
mu_w2 = 4.0           # ???
Zq_w2 = 1.005         # rms ionic charge
gamma_w2 = 1.01       # Leitherer 1995, ????
T_w2 = 0.3 * T_st2    # Drew1990 for r >> Rst
```

```python
def gaunt_ff(nu, T_w, Zq_w):
    return 9.77 * (1.0 + 0.13 * np.log10( np.power(T_w,1.5)/Zq_w/nu) ) 

def S_ff(nu, Mdot, vinf, T_w, mu_w, gamma_w, Zq_w, d):
    '''Free-free emission from the stellar wind (S_nu \propto nu^0.6) in Jy.
    Eq. 8 from https://articles.adsabs.harvard.edu/pdf/1975MNRAS.170...41W'''
    v5 = vinf/1e5 # cm/s --> km/s
    Mdot_eff = Mdot/np.sqrt(f_w)
    M_adim = Mdot_eff/M_sun_yr
    d_adim = d/kpc
    g_ff = gaunt_ff(nu, T_w, Zq_w)
    return (23.2/d_adim**2) * np.power( (M_adim/v5/mu_w)**2 *nu*gamma_w*g_ff*Zq_w**2, 2.0/3.0 ) 

def R_nu(nu, Mdot, vinf, T_w, mu_w, gamma_w, Zq_w, d):
    '''Characteristic radius of the bremsstrahlung emission (tau_nu = 1)
    Eq. 20 from Daley-Yates et al. 2016'''
    v5 = vinf/1e5 # cm/s --> km/s
    Mdot_eff = Mdot/np.sqrt(f_w)
    M_adim = Mdot_eff/M_sun_yr
    d_adim = d/kpc
    g_ff = gaunt_ff(nu, T_w, Zq_w)
    return 1.75e28 / np.sqrt(T_w) * np.cbrt( ( M_adim*Zq_w*g_ff/(mu_w*v5*nu) )**2 * gamma_w )

def nu_c(nu, Mdot, vinf, T_w, mu_w, gamma_w, Zq_w, R):
    '''Frequency at which the spectrum of the stellar wind flattens (S_nu \propto nu^-0.1).
    Eq. 12 from https://articles.adsabs.harvard.edu/pdf/1975MNRAS.170...41W'''
    v5 = vinf/1e5 # cm/s --> km/s
    Mdot_eff = Mdot/np.sqrt(f_w)
    M_adim = Mdot_eff/M_sun_yr
    g_ff = gaunt_ff(nu, T_w, Zq_w)
    return 4.7e42 * np.sqrt(gamma_w*g_ff) * Zq_w * M_adim / (np.power(T_w * R**2, 3.0/4.0 ) * mu_w * v5) # Hz
```

### Compare the expected free-free flux with the upper limit of ~50 uJy at 1 GHz

```python
nu = 1.0e9 # 1 GHz
S1 = S_ff(nu, Mdot1, vinf1, T_w1, mu_w1, gamma_w1, Zq_w1, d1 ) 
S2 = S_ff(nu, Mdot2, vinf2, T_w2, mu_w2, gamma_w2, Zq_w2, d2 ) 

#f'S1(WN) = {round(S1 * 1e6,1)} uJy'
print(f'S1(WN) = {S1 * 1e6 :.1f} uJy')
print(f'S2(WC) = {S2 * 1e6 :.1f} uJy')
```

### Calculate the fluxes and characteristic radii at the center of the uGMRT bands

```python
nu_GMRT = np.array([735, 1260])*1e6 # 735, 1260 MHz
S1_GMRT = S_ff(nu_GMRT, Mdot1, vinf1, T_w1, mu_w1, gamma_w1, Zq_w1, d1 ) 
S2_GMRT = S_ff(nu_GMRT, Mdot2, vinf2, T_w2, mu_w2, gamma_w2, Zq_w2, d2 ) 
R1_GMRT = R_nu(nu_GMRT, Mdot1, vinf1, T_w1, mu_w1, gamma_w1, Zq_w1, d1 ) 
R2_GMRT = R_nu(nu_GMRT, Mdot2, vinf2, T_w2, mu_w2, gamma_w2, Zq_w2, d2 ) 

for i in range(len(nu_GMRT)):
    print(f'{nu_GMRT[i] / 1e6 :.0f} MHz: R1[AU] = {R1_GMRT[i]/AU:.1f}, R2[AU] = {R2_GMRT[i]/AU:.1f}')
    
for i in range(len(nu_GMRT)):
    print(f'{nu_GMRT[i] / 1e6 :.0f} MHz: S1 = {S1_GMRT[i] * 1e6 :.1f} uJy, S2 = {S2_GMRT[i] * 1e6 :.1f} uJy')
```

### Calculate the frequency up to which the wind has a slope of 0.6

```python
nu = 1e12 # 1 THz, just a reference number? 
nu_c1 = nu_c(nu, Mdot1, vinf1, T_w1, mu_w1, gamma_w1, Zq_w1, Rst1 ) 
nu_c2 = nu_c(nu, Mdot2, vinf2, T_w2, mu_w2, gamma_w2, Zq_w2, Rst2 ) 

print(f'nu_c1(WN) = {nu_c1 / 1e12 :.1f} THz')
print(f'nu_c2(WC) = {nu_c2 / 1e12 :.1f} THz')
```
