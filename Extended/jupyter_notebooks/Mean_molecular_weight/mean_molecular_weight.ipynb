```python
import numpy as np
```

```python
# Solar abundance
X = 0.73
Y = 0.25
Z = 0.02

# WN star
#X = 0.1
#Y = 0.5
#Z = 1 - X - Y 

# WC star
#X = 0.001
#Y = 0.4
#Z = 1 - X - Y 

# For total ionization 
mu_t = 4 / ( 3 + 5*X - Z )
mu_e = 2 / ( X + 1 )
mu_i = 1. / (1./mu_t - 1./mu_e)

print('For fully ionised gas:')
print(f'{mu_t = :.2f}')
print(f'{mu_i = :.2f}')
print(f'{mu_e = :.2f}')


# For a neutral medium 
A_mean = 15.5 # for solar composition (http://astronomy.nmsu.edu/jasonj/565/docs/09_03.pdf)
Z_mean = A_mean/2    # for solar composition (I just considered A ~ 2*Z is valid also in the mean)
mu_neutro = 1. / ( X + Y/4 + Z/A_mean )

print('\nFor a neutral medium:')
print(f'{mu_neutro = :.2f}')


# Formulae from Padovani 2018 (https://www.aanda.org/articles/aa/pdf/2018/06/aa32202-17.pdf):
eps_pp = X + Y * 4**0.79 + Z * A_mean**0.79 # sum( x_i * A_i**0.79 ) , # Eq. A.3
eps_br = ( X * 2 + Y * 2 * 3 + Z * Z_mean*(Z_mean+1) ) / 2 # sum( x_i * Z_i*(Z_i+1) ) / 2 , # Eq. A.4

# Print the results
print(f'\nFor NT processes (no neutral H):')
print(f'{eps_pp = :.2f}')
print(f'{eps_br = :.2f}')
```

```python
# Define short names of the most relevant elements
#elements = ['H', 'He', 'C', 'N', 'O', 'Ne', 'Mg', 'Al', 'Si', 'S', 'Ar', 'Fe', 'Ni']

# Define relative abundances to Wilms+2000 (ISM)
#abundances = [0.835, 0.163, 4.01e-4, 1.27e-4, 8.19e-4, 1.46e-4, 4.19e-5, 3.57e-5, 3.11e-5, 2.05e-5, 4.29e-6, 4.49e-5, 1.87e-6]

#------------------------------------------------------------------------------


# Define short names of elements
elements = ['H', 'He', 'C', 'N', 'O', 'Ne', 'Mg', 'Si', 'S', 'Fe', 'Ni',
            'Ca', 'Al', 'Ar', 'Na', 'P', 'Cl', 'K', 'Ti', 'Cr']

# Define atomic masses (A, rounded to nearest integer)
mass_numbers = [1, 4, 12, 14, 16, 20, 24, 28, 32, 56, 58, 40, 27, 
                40, 23, 31, 35, 39, 48, 52]

# Define atomic numbers (Z)
atomic_numbers = [1, 2, 6, 7, 8, 10, 12, 14, 16, 26, 28, 20, 13, 
                  18, 11, 15, 17, 19, 22, 24]

# Define the abudances at solar values (from Chat GPT, use with caution!)
abundances = [0.739, 0.28, 2.92e-3, 1.01e-3, 8.51e-4, 1.26e-4, 7.10e-5, 
              3.55e-5, 1.63e-5, 3.16e-5, 1.49e-5, 2.29e-6, 2.01e-6, 
              2.24e-6, 2.10e-6, 8.30e-7, 5.15e-7, 4.62e-7, 4.00e-7, 3.60e-7]

# Normalize the abundances to sum up to 1 (due to rounding errors)
renorm = sum(abundances)
abundances = [abundance/renorm for abundance in abundances]

n_el = len(elements)

#-----------------------------------------------------------------------------

# Calculate the mean atomic weight in the gas (mu)
mu = sum([abundances[i] * mass_numbers[i] for i in range(n_el)])

# Calculate the Zrms
Zrms = np.sqrt( sum([abundances[i] * atomic_numbers[i]**2 for i in range(n_el)]) )

# For fully ionised gas
mu_ionised = sum([abundances[i] * mass_numbers[i]/(1+atomic_numbers[i]) for i in range(n_el)])
mu_e = 2 / ( X + 1 )
mu_i = 1. / (1./mu_t - 1./mu_e)


# Print the results
print('For neutral gas:')
print(f'{mu = :.2f}')
print(f'{Zrms = :.2f}')

print('\nFor fully ionised gas:')
print(f'{mu_ionised = :.2f}')
print(f'{mu_i = :.2f}')
print(f'{mu_e = :.2f}')


# Formulae from Padovani 2018 (https://www.aanda.org/articles/aa/pdf/2018/06/aa32202-17.pdf):
eps_pp = sum([abundances[i] * mass_numbers[i]**0.79 for i in range(n_el)]) # Eq. A.3
eps_br = sum([abundances[i] * atomic_numbers[i]*(atomic_numbers[i]+1) / 2 for i in range(n_el)]) # Eq. A.4

# Print the results
print(f'\nFor NT processes (no neutral H):')
print(f'{eps_pp = :.2f}')
print(f'{eps_br = :.2f}')
```
