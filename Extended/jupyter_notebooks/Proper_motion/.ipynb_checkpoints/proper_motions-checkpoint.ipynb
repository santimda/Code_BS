## This notebook is used to: 
### i) transform proper motions to Galactic coordinates
### ii) correct for Galactic rotation 
### iii) convert back to equatorial motions with respect to the surrounding medium.

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.coordinates as coord
import astropy.units as u
from uncertainties import ufloat
from uncertainties.umath import *
r_a = ufloat(308.7296157020184, 0.010217067)
decl = ufloat(40.57895237587182, 0.010932052)
parallax = ufloat(0.5834730765997992, 0.012209491)
pmra = ufloat(-2.594735635507606,0.013595376)
pmdec = ufloat(0.7286401164024895,0.014304411)

ra=r_a*3
print(ra)
```

### First we define some general functions

```python
def galactic_velocity(dist, b, l, A, B, C, K, U, V, W):
    '''Galactic velocity curve from Wang et al. 2021 (https://arxiv.org/pdf/2103.10232.pdf, Eqs.7-9). 
    Arguments:
    dist: distance in kpc
    b, l: galactic coordinates in radians
    A, B, C, K: Oort constants in km/s/kpc
    U, V, W: Solar velocity in km/s'''

    cb = np.cos(b)
    sb = np.sin(b)
    cl = np.cos(l)
    sl = np.sin(l)
    c2l = np.cos(2*l)
    s2l = np.sin(2*l)
    d_adim = dist/u.kpc
    
    V_r0 = -U*cl*cb - V*sl*cb - W*sb + d_adim * (K + A*s2l + C*c2l) * cb*cb 
                    
    V_l0 = U*sl - V*cl + d_adim * (A*c2l + B - C*s2l) * cb
    
    V_b0 = U*cl*sb + V*sl*sb - W*cb - d_adim * (K + A*s2l + C*c2l) * sb*cb  
    
    return V_r0, V_l0, V_b0


def velocity_to_propermotion(dist, V_l0, V_b0):
    '''Transform V_l0 and V_b0 to mu_l0 and mu_b0.
    dist: distance in kpc
    V_l0, V_b0: velocities in km/s'''
    
    mu_l0 = 0.211/(dist/u.kpc) * V_l0 * u.mas/u.yr
    mu_b0 = 0.211/(dist/u.kpc) * V_b0 * u.mas/u.yr
    
    return mu_l0, mu_b0


def oort_constants(model):
    '''Retrieve the Oort constants for one of different models:
    C07: Comerón & Pascquali 2007
    B19: Bobylev & Bajkova 2019 (OSCs)
    W21: Wang et al. 2021 (https://arxiv.org/pdf/2103.10232.pdf)'''
    
    if model == 'C07':
        A = 12.5
        B = -12.5
        C = 0.0
        K = 0.0
        U = 7.0
        V = 14.0
        W = 7.0
    elif model == 'B19':
        A = 16.40
        B = -12.31
        C = 0.0
        K = 0.0
        U = 8.53
        V = 11.22
        W = 7.83
    elif model == 'W21':
        A = 16.31
        B = -11.99
        C = -3.10
        K = -1.25
        U = 11.69
        V = 10.16
        W = 7.67
    else:
        print('ERROR! wrong model')
 
    return A, B, C, K, U, V, W
```

### Second we introduce the barycentric coordinates and velocity in the ICRS for the object

```python
# Gaia DR3 data for BD+43 3654 (V_r from Kobulnicky2010)
r_a = ufloat(308.7296157020184, 0.010217067)
decl = ufloat(40.57895237587182, 0.010932052)
parallax = ufloat(0.5834730765997992, 0.012209491)
pmra = ufloat(-2.594735635507606,0.013595376)
pmdec = ufloat(0.7286401164024895,0.014304411)

BD = coord.SkyCoord(ra=r_a*u.degree, dec=decl*u.degree,
                    distance=(parallax*u.mas).to(u.kpc, u.parallax()),
                    pm_ra_cosdec=pmra*u.mas/u.yr,
                    pm_dec=pmdec*u.mas/u.yr,
                    radial_velocity=10*u.km/u.s, 
                    frame='icrs')

# Print the values to check they are OK
#print(BD.galactic)

# Save in shorter-name variables
l, b, D = BD.galactic.l, BD.galactic.b, BD.galactic.distance
mu_l, mu_b = BD.galactic.pm_l_cosb, BD.galactic.pm_b
V_r = BD.galactic.radial_velocity
print(f'{D =}')
print(f'{l =}, {b=}')
print(f'{mu_l =}, {mu_b=}')
print(f'{V_r =}')
```

### Third, we calculate local Galactic velocity field and the object proper motion w.r.t. local medium

```python
A, B, C, K, U, V, W = oort_constants('C07')
#A, B, C, K, U, V, W = oort_constants('B19')
#A, B, C, K, U, V, W = oort_constants('W21')


V_r0, V_l0, V_b0 = galactic_velocity(D, b, l, A, B, C, K, U, V, W)
mu_l0, mu_b0 = velocity_to_propermotion(D, V_l0, V_b0)

print(' Galactic rotation in mas/yr:')
print('(\u03BC_l)0=' + str(round(mu_l0.value, 2)) + ', (\u03BC_b)0=' + str(round(mu_b0.value, 2)) )

# Proper motions w.r.t. the surrounding medium:
mu_l_corr = mu_l - mu_l0
mu_b_corr = mu_b - mu_b0

print('\n Corrected values:')
print('\u03BC_l=' + str(round(mu_l_corr.value, 2)) + ', \u03BC_b=' + str(round(mu_b_corr.value, 2)) )

#Calculate mu_alpha, mu_dec, and V_tan
BD_corr = coord.SkyCoord(l=l, b=b,
                    distance=D,
                    pm_l_cosb=mu_l_corr,
                    pm_b=mu_b_corr,
                    frame='galactic')

icrs = BD_corr.icrs
print(f'\n {icrs=}')
angle = np.arctan(icrs.pm_dec/icrs.pm_ra_cosdec).to(u.deg) 
print( '\n angle [deg] =', str(round(angle.value, 2 )) )

# Vl= 4.74*r*μ_l_cosb
# Vb= 4.74*r*μ_b
# V_t = 4.74*D*sqrt(μ_l_cosb^2 + μ_b^2)
V_t = 4.74*u.yr/u.mas*D/u.kpc * np.sqrt(mu_l_corr**2 + mu_b_corr**2)*u.km/u.s
print( '\n V_t [km/s]=', str(round(V_t.value, 2 )) ) 

V_r_corr = V_r - V_r0*u.km/u.s
print( '\n V_r [km/s]=', str(round(V_r_corr.value, 2 )) ) 
print( '\n v_tot [km/s]=', str(round(np.sqrt(V_t.value**2+V_r_corr.value**2), 2)) ) 
```
