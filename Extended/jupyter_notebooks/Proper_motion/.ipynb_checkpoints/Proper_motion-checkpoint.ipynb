## This notebook is used to: i) transform proper motions to Galactic coordinates, ii) correct for Galactic rotation, iii) convert back to equatorial motions with respect to the surrounding medium.

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.coordinates as coord
import astropy.units as u
from astropy.visualization import astropy_mpl_style
plt.style.use(astropy_mpl_style)
```

### First we define define a barycentric coordinate and velocity in the ICRS frame using the Gaia DR3 data for BD+43 3654 (V_r from Kobulnicky2010). 

```python
# Gaia DR3
BD = coord.SkyCoord(ra=350.1854483354341*u.degree, dec=61.19459355028689*u.degree,
                    distance=(0.33406384982164944*u.mas).to(u.kpc, u.parallax()),
                    pm_ra_cosdec=-2.662478918588872*u.mas/u.yr,
                    pm_dec=8.113332192336172*u.mas/u.yr,
                    radial_velocity=20*u.km/u.s, 
                    frame='icrs')

# Comerón & Pasquali 2007
#BD = coord.SkyCoord(ra=308.40031523381674*u.degree, dec=43.98538212854551*u.degree,
#                    distance=1.450*u.kpc,
#                    pm_ra_cosdec=-0.4*u.mas/u.yr,
#                    pm_dec=1.3*u.mas/u.yr,
#                    radial_velocity=-66.2*u.km/u.s, 
#                    frame='icrs')


# Print the values to check they are OK
BD.galactic
```

```python
# Save in shorter-name variables
l, b, D = BD.galactic.l, BD.galactic.b, BD.galactic.distance
mu_l, mu_b = BD.galactic.pm_l_cosb, BD.galactic.pm_b
print(l, b, D)
print(mu_l, mu_b)
```

### Define Oort constants, calculate local Galactic velocity field and proper motion w.r.t. local medium 
(use Comerón & Pasquali 2007 values)

```python
# Constants from Comerón & Pascquali 2007 [km/s/kpc]
A = 12.5
B = -12.5
U = 7.0
V = 14.0
W = 7.0

# Calculate the local Galactic velocity field (Comerón & Pascquali 2007, Eq.2)
mu_l0 = 0.211*( A*np.cos(2*l)*np.cos(b) 
               + B*np.cos(b) 
               + (U*u.kpc/D)*np.sin(l) 
               - (V*u.kpc/D)*np.cos(l) 
              )*u.mas/u.yr

mu_b0 = 0.211*( -A*np.sin(2*l)*np.sin(b)*np.cos(b) 
               + (U*u.kpc/D)*np.cos(l)*np.sin(b) 
               + (V*u.kpc/D)*np.sin(l)*np.sin(b) 
               - (W*u.kpc/D)*np.cos(b) 
              )*u.mas/u.yr

print('Rotación galáctica: (\u03BC_l)0=' + str(mu_l0) + ', (\u03BC_b)0=' + str(mu_b0) )

# Proper motions w.r.t. the surrounding medium:
mu_l_corr = mu_l - mu_l0
mu_b_corr = mu_b - mu_b0

print('Valores corregidos: \u03BC_l=' + str(mu_l_corr) + ', \u03BC_b=' + str(mu_b_corr) )
```

Calculate mu_alpha, mu_dec, and V_tan

```python
BD_corr = coord.SkyCoord(l=l, b=b,
                    distance=D,
                    pm_l_cosb=mu_l_corr,
                    pm_b=mu_b_corr,
                    frame='galactic')

icrs = BD_corr.icrs
print(icrs)
angle = np.arctan(icrs.pm_dec/icrs.pm_ra_cosdec).to(u.deg) 
print( '\n angle [deg] =', str(round(angle.value, 2 )) )

# Vl= 4.74*r*μ_l_cosb
# Vb= 4.74*r*μ_b
# V_t = 4.74*D*sqrt(μ_l_cosb^2 + μ_b^2)
V_t = 4.74*u.yr/u.mas*D/u.kpc*np.sqrt(mu_l_corr**2 + mu_b_corr**2)*u.km/u.s
print( '\n V_tan [km/s]=', str(round(V_t.value, 2 )) ) 
```

### Define Oort constants, calculate local Galactic velocity field and proper motion w.r.t. local medium 
(use Bobylev and Bajkova 2019 values)

```python
# Constants from Bobylev & Bajkova 2019 (OSCs)
A = 16.40
B = -12.31
U = 8.53
V = 11.22
W = 7.83

# Calculate the local Galactic velocity field (Comerón & Pascquali 2007, Eq.2)
mu_l0 = 0.211*( A*np.cos(2*l)*np.cos(b) 
               + B*np.cos(b) 
               + U*u.kpc/D*np.sin(l) 
               - V*u.kpc/D*np.cos(l) )*u.mas/u.yr

mu_b0 = 0.211*( -A*np.sin(2*l)*np.sin(b)*np.cos(b) 
               + U*u.kpc/D*np.cos(l)*np.sin(l) 
               - V*u.kpc/D*np.sin(l)*np.sin(b) 
               - W*u.kpc/D*np.cos(b) )*u.mas/u.yr

print('Rotación galáctica: (\u03BC_l)0=' + str(mu_l0) + ', (\u03BC_b)0=' + str(mu_b0) )

# Proper motions w.r.t. the surrounding medium:
mu_l_corr = mu_l - mu_l0
mu_b_corr = mu_b - mu_b0

print('Valores corregidos: \u03BC_l=' + str(mu_l_corr) + ', \u03BC_b=' + str(mu_b_corr) )
```

Calculate mu_alpha, mu_dec, and V_tan

```python
BD_corr = coord.SkyCoord(l=l, b=b,
                    distance=D,
                    pm_l_cosb=mu_l_corr,
                    pm_b=mu_b_corr,
                    frame='galactic')

icrs = BD_corr.icrs
print(icrs)
angle = np.arctan(icrs.pm_dec/icrs.pm_ra_cosdec).to(u.deg) 
print( '\n angle [deg] =', str(round(angle.value, 2 )) )

# Vl= 4.74*r*μ_l_cosb
# Vb= 4.74*r*μ_b
# V_t = 4.74*D*sqrt(μ_l_cosb^2 + μ_b^2)
V_t = 4.74*u.yr/u.mas*D/u.kpc*np.sqrt(mu_l_corr**2 + mu_b_corr**2)*u.km/u.s
print( '\n V_tan [km/s]=', str(round(V_t.value, 2 )) ) 
```
